```python
employee = [
    {"id": 1, "name": "Rahul", "salary": 50000, "location": "Mumbai"},
    {"id": 2, "name": "Priya", "salary": 60000, "location": "Pune"},
    {"id": 3, "name": "Amit", "salary": 55000, "location": "Delhi"},
    {"id": 4, "name": "Sneha", "salary": 65000, "location": "Bangalore"}
]
```

```python
attendance = [
    # Rahul
    {"empId": 1, "date": "2026-09-20", "status": "Present"},
    {"empId": 1, "date": "2026-09-21", "status": "Present"},
    {"empId": 1, "date": "2026-09-22", "status": "Absent"},
    {"empId": 1, "date": "2026-09-23", "status": "Present"},
    {"empId": 1, "date": "2026-09-24", "status": "Present"},

    # Priya
    {"empId": 2, "date": "2026-09-20", "status": "Present"},
    {"empId": 2, "date": "2026-09-21", "status": "Absent"},
    {"empId": 2, "date": "2026-09-22", "status": "Present"},
    {"empId": 2, "date": "2026-09-23", "status": "Present"},
    {"empId": 2, "date": "2026-09-24", "status": "Absent"},

    # Amit
    {"empId": 3, "date": "2026-09-20", "status": "Present"},
    {"empId": 3, "date": "2026-09-21", "status": "Present"},
    {"empId": 3, "date": "2026-09-22", "status": "Present"},
    {"empId": 3, "date": "2026-09-23", "status": "Absent"},
    {"empId": 3, "date": "2026-09-24", "status": "Present"},

    # Sneha
    {"empId": 4, "date": "2026-09-20", "status": "Absent"},
    {"empId": 4, "date": "2026-09-21", "status": "Present"},
    {"empId": 4, "date": "2026-09-22", "status": "Present"},
    {"empId": 4, "date": "2026-09-23", "status": "Present"},
    {"empId": 4, "date": "2026-09-24", "status": "Present"}
]
```

```python
result = []

for emp in employee:
    records = [a for a in attendance if a["empId"]==emp["id"]]

    result.append({
        "id":emp["id"],
        "name":emp["name"],
        "Present":sum(a["status"]=="Present" for a in records),
        "Absent":sum(a["status"]=="Absent" for a in records)
    })

result
```
