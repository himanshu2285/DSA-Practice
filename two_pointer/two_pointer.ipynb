### Move Zeros

```python
from typing import List
class Solution:
    def two_sum(self, nums: List[int]) -> None:
        i = 0
        for j in range(len(nums)):
            if nums[j]!=0:
                nums[i], nums[j] = nums[j], nums[i]
                i+=1
```

### Valid Palindrome

```python
from typing import List
import re
class Solution:
    def palindrome(self, s: List[int]) -> bool:
        if s == " ":
            return True
        else:
            # re.sub(pattern, replacement, string)
            cleaned_data = re.sub(r'[^a-zA-Z0-9]', '', s)
            if cleaned_data.lower()==cleaned_data[::-1].lower():
                return True
            return False

obj = Solution()
# print(obj.palindrome("hey there, hhow are you"))
print(obj.palindrome("navan"))
```

### Two Sum II

```python
from typing import List
class Solution:
    def twosum2(self, numbers: List[int], target):
        left = 0
        right = len(numbers)-1
        while left<right:
            curr_sum = numbers[left] + numbers[right]
            if curr_sum == target:
                return [left+1, right+1]
            elif curr_sum > target:
                right -= 1
            else:
                left += 1

obj = Solution()
print(obj.twosum2([2,7,11,15], 9))
print(obj.twosum2([2,3,4], 6))
print(obj.twosum2([-1,0], -1))
```

### 11. Container With Most Water

```python
class Solution:
    def maxArea(self, height: List[int]) -> int:
        left = 0
        right = len(height) - 1
        max_water = 0
        
        while left < right:
            width = right - left
            
            curr_water = width * min(height[left], height[right])
            
            max_water = max(curr_water, max_water)
            
            if height[left] < height[right]:
                left += 1
            else:
                right -= 1
        return max_water
    
obj = Solution()
print(obj.maxArea([1,8,6,2,5,4,8,3,7]))        
print(obj.maxArea([1,1]))        
```
